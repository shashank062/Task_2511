```python
def evaluate_trade(trade):

    portfolio = portfolios[trade["portfolio_id"]]

    security_id = trade["security_id"]
    action = trade["action"]
    quantity = trade["quantity"]
    price = trade["price"]

    if quantity <= 0 or price <= 0:
        return "Rejected", "Invalid trade data."

    if action not in ["Buy", "Sell"]:
        return "Rejected", "Invalid trade data."

    if security_id in portfolio["restricted_securities"]:
        return (
            "Rejected",
            "Security is restricted for this portfolio."
        )


    if action == "Sell":

        holdings = portfolio["current_holdings"]

        current_qty = holdings.get(
            security_id,
            {"quantity": 0}
        )["quantity"]

        if current_qty < quantity:
            return (
                "Rejected",
                "Insufficient holdings to sell."
            )

        return (
            "Approved",
            "Trade passed all checks."
        )


    cost = quantity * price

    if cost > portfolio["cash_balance"]:
        return (
            "Rejected",
            "Insufficient cash for this buy order."
        )


    holdings = portfolio["current_holdings"]

    existing_qty = holdings.get(
        security_id,
        {"quantity": 0}
    )["quantity"]

    post_trade_cash = (
        portfolio["cash_balance"] - cost
    )

    post_trade_quantity = (
        existing_qty + quantity
    )

    post_trade_security_value = (
        post_trade_quantity * price
    )

    other_holdings_value = 0

    for sec, data in holdings.items():

        if sec != security_id:

            other_holdings_value += (
                data["quantity"]
                * data["market_price"]
            )

    portfolio_value_after = (
        post_trade_cash
        + post_trade_security_value
        + other_holdings_value
    )

    position_pct = (
        post_trade_security_value
        / portfolio_value_after
    )

    if (
        position_pct
        > portfolio["max_position_pct"]
    ):
        return (
            "Needs Manual Sign-off",
            "Exceeds maximum position concentration — requires approval."
        )

    return (
        "Approved",
        "Trade passed all checks."
    )
```

```python
test_trade = {
    "portfolio_id": "POR001",
    "security_id": "AAPL",
    "action": "Buy",
    "quantity": 20,
    "price": 150,
    "security_type": "Equity"
}

evaluate_trade(test_trade)
```

```python
results = []

for _, trade in trades.iterrows():

    decision, reason = evaluate_trade(trade)

    results.append([
        trade["portfolio_id"],
        trade["security_id"],
        trade["action"],
        decision,
        reason
    ])
```

```python
output = pd.DataFrame(
    results,
    columns=[
        "portfolio_id",
        "security_id",
        "action",
        "decision",
        "reason"
    ]
)

output
```

```python
output.to_csv(
    "trade_results.csv",
    index=False
)

print("Results saved successfully!")
```
